```python
import pandas as pd
import numpy as np
```

```python
l_num = np.arange(0, 10).tolist()
```

```python
def leer_csv(ruta: str) -> pd.DataFrame:
    df = pd.read_csv(ruta, sep=';', encoding='latin-1', dtype='str')
    df.columns = (df.columns.str.strip()
                  .str.lower()
                  .str.replace(' ', '_')
                  .str.replace('ó', 'o')
                  .str.replace('í', 'i')
                 )
    return df
```

```python
stock_2019 = leer_csv('../data/raw/stock-2019.csv')
nacimientos_2019 = leer_csv('../data/raw/nacimientos-2019.csv')
muertes_2019 = leer_csv('../data/raw/muertes-2019.csv')

stock_2020 = leer_csv('../data/raw/stock-2020.csv')
nacimientos_2020 = leer_csv('../data/raw/nacimientos-2020.csv')
muertes_2020 = leer_csv('../data/raw/muertes-2020.csv')

stock_2021 = leer_csv('../data/raw/stock-2021.csv')
nacimientos_2021 = leer_csv('../data/raw/nacimientos-2021.csv')
muertes_2021 = leer_csv('../data/raw/muertes-2021.csv')

stock_2022 = leer_csv('../data/raw/stock-2022.csv')
nacimientos_2022 = leer_csv('../data/raw/nacimientos-2022.csv')
muertes_2022 = leer_csv('../data/raw/muertes-2022.csv')
```

```python
def numero(s:pd.Series) -> pd.Series:
    return pd.to_numeric(s.str.strip().str.replace('.', '')
                         , errors='coerce')
```

```python
nacimientos_2019['total'] = numero(nacimientos_2019['total'])
muertes_2019['total'] = numero(muertes_2019['total'])
stock_2019['total'] = numero(stock_2019['total'])

nacimientos_2020['total'] = numero(nacimientos_2020['total'])
muertes_2020['total'] = numero(muertes_2020['total'])
stock_2020['total'] = numero(stock_2020['total'])

nacimientos_2021['total'] = numero(nacimientos_2021['total'])
muertes_2021['total'] = numero(muertes_2021['total'])
stock_2021['total'] = numero(stock_2021['total'])

nacimientos_2022['total'] = numero(nacimientos_2022['total'])
muertes_2022['total'] = numero(muertes_2022['total'])
stock_2022['total'] = numero(stock_2022['total'])
```

```python
# filtrado
def por_actividad(df:pd.DataFrame) -> pd.Series:
    return df[(df['actividad_principal_(cnae_2009)'] != 'Todas las actividades') & (df['condicion_juridica'] == 'Personas físicas') & (df['comunidades_y_ciudades_autonomas'].isna())].set_index('actividad_principal_(cnae_2009)')['total']
```

```python
dnueve = pd.DataFrame({
    'stock': por_actividad(stock_2019),
    'nacimientos': por_actividad(nacimientos_2019),
    'muertes': por_actividad(muertes_2019),
})

veinte = pd.DataFrame({
    'stock': por_actividad(stock_2020),
    'nacimientos': por_actividad(nacimientos_2020),
    'muertes': por_actividad(muertes_2020),
})

vuno = pd.DataFrame({
    'stock': por_actividad(stock_2021),
    'nacimientos': por_actividad(nacimientos_2021),
    'muertes': por_actividad(muertes_2021),
})

vdos = pd.DataFrame({
    'stock': por_actividad(stock_2022),
    'nacimientos': por_actividad(nacimientos_2022),
    'muertes': por_actividad(muertes_2022),
})
```

```python
dnueve['t_nacimientos'] = np.round((dnueve['nacimientos'] / dnueve['stock']) * 100, 2)
dnueve['t_muertes'] = np.round((dnueve['muertes'] / dnueve['stock']) * 100, 2)
dnueve ['t_neta'] = dnueve['t_nacimientos'] - dnueve['t_muertes']

veinte['t_nacimientos'] = np.round((veinte['nacimientos'] / veinte['stock']) * 100, 2)
veinte['t_muertes'] = np.round((veinte['muertes'] / veinte['stock']) * 100, 2)
veinte ['t_neta'] = veinte['t_nacimientos'] - veinte['t_muertes']

vuno['t_nacimientos'] = np.round((vuno['nacimientos'] / vuno['stock']) * 100, 2)
vuno['t_muertes'] = np.round((vuno['muertes'] / vuno['stock']) * 100, 2)
vuno ['t_neta'] = vuno['t_nacimientos'] - vuno['t_muertes']

vdos['t_nacimientos'] = np.round((vdos['nacimientos'] / vdos['stock']) * 100, 2)
vdos['t_muertes'] = np.round((vdos['muertes'] / vdos['stock']) * 100, 2)
vdos ['t_neta'] = vdos['t_nacimientos'] - vdos['t_muertes']
```

```python
unido = pd.DataFrame({
    '2019': dnueve['t_neta']
    ,'2020': veinte['t_neta']
    ,'2021': vuno['t_neta']
    ,'2022': vdos['t_neta']
})
    #.set_index('actividad_principal_(cnae_2009)')
```
